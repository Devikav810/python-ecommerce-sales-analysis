```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
print("Rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

df.head()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

df.info()
```

```python
df.describe(include="all").T
```

```python
df = df.drop_duplicates().copy()

df.columns = df.columns.str.strip().str.replace(
    r"\s+", "_", regex=True
)

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"], errors="coerce"
)

numeric_cols = [
    "Quantity", "Unit_Price", "Discount_Percent",
    "Total_Sales", "Shipping_Cost", "Profit"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(
    subset=["Order_ID", "Order_Date", "Quantity", "Unit_Price"]
).copy()

df = df[
    (df["Quantity"] > 0) &
    (df["Unit_Price"] >= 0)
].copy()

for col in ["Discount_Percent", "Total_Sales", "Shipping_Cost", "Profit"]:
    df[col] = df[col].fillna(df[col].median())

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].replace("", np.nan)
    df[col] = df[col].fillna("Unknown")

df["Discount_Percent"] = df["Discount_Percent"].clip(0, 100)

print("Data cleaning completed!")
print("Cleaned shape:", df.shape)
```

```python
df["Gross_Sales"] = df["Quantity"] * df["Unit_Price"]

df["Discount_Amount"] = (
    df["Gross_Sales"] - df["Total_Sales"]
).clip(lower=0)

df["Profit_Margin_Percent"] = np.where(
    df["Total_Sales"] > 0,
    df["Profit"] / df["Total_Sales"] * 100,
    np.nan
)

df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Month_Name"] = df["Order_Date"].dt.strftime("%B")
df["Quarter"] = df["Order_Date"].dt.quarter
df["Day_Name"] = df["Order_Date"].dt.day_name()

df["Order_Size"] = np.where(
    df["Total_Sales"] >= df["Total_Sales"].median(),
    "High Value",
    "Low Value"
)

df.head()
```

```python
print("Total Sales:", round(df["Total_Sales"].sum(), 2))
print("Total Profit:", round(df["Profit"].sum(), 2))
print("Total Quantity:", df["Quantity"].sum())
print("Number of Records:", len(df))
print("Average Sales:", round(df["Total_Sales"].mean(), 2))
```

```python
category_sales = df.groupby(
    "Product_Category"
)["Total_Sales"].sum().sort_values(ascending=False)

display(category_sales)
```

```python
country_sales = df.groupby(
    "Country"
)["Total_Sales"].sum().sort_values(ascending=False)

display(country_sales.head(10))
```

```python
monthly_sales = df.groupby(
    df["Order_Date"].dt.to_period("M")
)["Total_Sales"].sum()

display(monthly_sales)
```

```python
category_sales.plot(kind="bar", figsize=(10, 5))

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
country_sales.head(10).plot(
    kind="bar", figsize=(10, 5)
)

plt.title("Top 10 Countries by Sales")
plt.xlabel("Country")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly_sales.index = monthly_sales.index.astype(str)

plt.figure(figsize=(12, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Total_Sales"], bins=20, kde=True)

plt.title("Distribution of Sales")
plt.xlabel("Total Sales")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.boxplot(data=df, x="Order_Size", y="Total_Sales")

plt.title("Sales by Order Size")
plt.xlabel("Order Size")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```
